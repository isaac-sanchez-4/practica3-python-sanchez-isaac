## Ejercicio 5

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):

    importe_bruto = unidades * precio_unitario
    importe_neto = importe_bruto * (1 - descuento_pct / 100)

    return round(importe_neto, 2)


def resumen_pedido(unidades, precio_unitario, descuento_pct=0):

    importe_bruto = unidades * precio_unitario
    ahorro = importe_bruto * descuento_pct / 100
    importe_neto = calcular_importe(unidades, precio_unitario, descuento_pct)

    return (
        round(importe_bruto, 2),
        round(ahorro, 2),
        importe_neto
    )


importe_bruto, ahorro, importe_neto = resumen_pedido(
    unidades=3,
    precio_unitario=59.90,
    descuento_pct=10
)

print("Importe bruto:", importe_bruto, "€")
print("Ahorro:", ahorro, "€")
print("Importe neto:", importe_neto, "€")

#yuda de la función
help(calcular_importe)


pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = []

for pedido in pedidos:
    importe = calcular_importe(*pedido)
    importes.append(importe)

print("\nImportes:", importes)

print("Importe máximo:", max(importes))
print("Importe mínimo:", min(importes))
print("Importes ordenados:", sorted(importes))
print("Número de importes:", len(importes))

print(
    "\nComprobación:",
    calcular_importe(3, 59.90, 10)
)
```

## Ejercicio 11

```python
import numpy as np

#generar datos reproducibles
rng = np.random.default_rng(2024)

temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

print("Número de temperaturas:", len(temperatura))
print("Número de vibraciones:", len(vibracion))


#Estadística  de temperatura
media = np.mean(temperatura)
mediana = np.median(temperatura)
desviacion = np.std(temperatura)
minimo = np.min(temperatura)
maximo = np.max(temperatura)

print("\nESTADÍSTICA DE TEMPERATURA")
print("Media:", round(media, 2))
print("Mediana:", round(mediana, 2))
print("Desviación típica:", round(desviacion, 2))
print("Mínimo:", round(minimo, 2))
print("Máximo:", round(maximo, 2))

# La media y la mediana son muy próximas por lo que la distribucion de las temperatirads no son muy asimetricas


#Percentiles de vibración
percentiles = np.percentile(vibracion, [25, 50, 75, 95])

print("\nPERCENTILES DE VIBRACIÓN")
print("Percentil 25:", round(percentiles[0], 2))
print("Percentil 50:", round(percentiles[1], 2))
print("Percentil 75:", round(percentiles[2], 2))
print("Percentil 95:", round(percentiles[3], 2))


#matriz de temperatura y vibración
telemetria = np.column_stack((temperatura, vibracion))

print("\nMatriz:")
print("Shape:", telemetria.shape)


#matriz de correlacion
correlacion = np.corrcoef(temperatura, vibracion)

print("\nMatriz de correlación:")
print(correlacion)

print("\nCorrelación entre temperatura y vibración:",
      round(correlacion[0, 1], 3))

#La correlación permite medir la intensidad de la relación lineal entre las dos variables; un valor cercano a 0 indica una relación lineal débil


# unblad de temperatura
umbral = media + 2 * desviacion

lecturas_superan = np.sum(temperatura > umbral)
porcentaje = (lecturas_superan / len(temperatura)) * 100

print("Unbrales de temperatura")
print("Umbral:", round(umbral, 2), "°C")
print("Lecturas que superan el umbral:", lecturas_superan)
print("Porcentaje:", round(porcentaje, 2), "%")

# mediante la semilla 2024 podemos generar siempre los mismos datos aleatorios para poder reproducir el analisis y comprobar los resultados
```

```python
#Crea la columna importe_bruto = unidades * precio_unitario.
ventas["importe_bruto"] = ventas["unidades"] * ventas["precio_unitario"]

print("Importe bruto:")
print(ventas[["unidades", "precio_unitario", "importe_bruto"]].head())


# Crea la columna importe_neto aplicando descuento_pct, redondeada a dos decimales.
ventas["importe_neto"] = (
    ventas["importe_bruto"] *
    (1 - ventas["descuento_pct"] / 100)
).round(2)

print("\nImporte neto:")
print(
    ventas[
        ["importe_bruto", "descuento_pct", "importe_neto"]
    ].head()
)


#Crea la columna ticket_medio = importe_neto / unidades.
ventas["ticket_medio"] = (
    ventas["importe_neto"] / ventas["unidades"]
).round(2)

print("\nTicket medio:")
print(ventas[["importe_neto", "unidades", "ticket_medio"]].head())


#Agrupa por region y calcula la suma de importe_neto, ordenada de mayor a menor.
facturacion_region = (
    ventas.groupby("region")["importe_neto"]
    .sum()
    .sort_values(ascending=False)
)

print("Facturación neta por región:")
print(facturacion_region)


#Agrupa por categoria y canal a la vez, y calcula importe_neto medio y unidades totales usando .agg() con un diccionario de funciones.
resumen_categoria_canal = (
    ventas.groupby(["categoria", "canal"])
    .agg({
        "importe_neto": "mean",
        "unidades": "sum"
    })
)

print("Facturación media y unidades totales por categoría y canal:")
print(resumen_categoria_canal)


# Obtén el top 5 de productos por facturación neta con .groupby(), .sum() y .nlargest().
top_5_productos = (
    ventas.groupby("producto")["importe_neto"]
    .sum()
    .nlargest(5)
)

print("Top 5 productos por facturación neta:")
print(top_5_productos)


#Rellena los nulos de satisfaccion con la mediana de la columna y los de canal con el texto "Desconocido".

ventas["satisfaccion"] = ventas["satisfaccion"].fillna(
    ventas["satisfaccion"].median()
)

ventas["canal"] = ventas["canal"].fillna("Desconocido")

print("\nValores nulos después de rellenarlos:")
print(ventas[["satisfaccion", "canal"]].isna().sum())


#Exporta el resultado agrupado por región a ../outputs/facturacion_por_region.csv.

facturacion_region.to_csv(
    "../outputs/facturacion_por_region.csv",
    header=["importe_neto"]
)

print("\nArchivo exportado correctamente:")
print("../outputs/facturacion_por_region.csv")
```
